3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
df[['systolic', 'distolic']] = df['Blood Pressure (s/d)'].str.split('/', expand=True).astype(int)
df['Stress_Group'] = pd.cut(df['Stress Levels'], bins=[0, 4, 7, 11], labels=['Low', 'Moderate', 'High'], include_lowest=True)
means = df.groupby('Stress_Group')['systolic'].mean()

diff = means['High'] - means['Low']
print(f"\nРазница (High - Low): {diff:.2f} mmHg")
print(f"Рост давления составляет: {(diff / means['Low'] * 100):.1f}%")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
df_women = df[df['Gender'] == 'Female'].copy()
df_women['Age_group'] = pd.cut(
    df_women['Age (years)'],
    bins=[0, 30, 59, 120],
    labels=['Young', 'Middle', 'Old'],
    include_lowest=True
)

df_filtered = df_women[df_women['Age_group'].isin(['Young', 'Old'])]
means = df_filtered.groupby('Age_group')['Cognitive Function'].mean()
diff = means['Old'] - means['Young']
percent_change = (diff / means['Young']) * 100


print(f"\nСредняя оценка Young: {means['Young']:.2f}")
print(f"Средняя оценка Old:   {means['Old']:.2f}")
````

**да, правда**

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
df_over_50 = df[df['Age (years)'] > 50].copy()

df_over_50['Sport'] = df_over_50['Physical Activity Level'].apply(
    lambda x: 'High Level' if x in ['High'] else 'Low Level'
)

mean_weights = df_over_50.groupby('Sport')['Weight (kg)'].mean()
print(f"Не занимаются спортом: {mean_weights.get('Low Level', 0):.2f} кг")
print(f"Занимаются спортом:    {mean_weights.get('High Level', 0):.2f} кг")

print(mean_weights.get('Low Level', 0) - mean_weights.get('High Level', 0))
````

**да, влияет. Люди с высокой активностью весят на 2.5 кг меньше**

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
import pandas as pd
import numpy as np

df['BMI_group'] = pd.cut(
    df['BMI'],
    bins=[0, 18.5, 25, 30, 50],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese'],
    include_lowest=True
)

print("Распределение ИМТ:")
print(df['BMI_group'].value_counts())

df['has_chronic_disease'] = df['Chronic Diseases'].apply(
    lambda x: 1 if pd.notna(x) and x != 'None' and x != '' else 0
)

chronic_disease_rate = df.groupby('BMI_group')['has_chronic_disease'].mean() * 100
print("Процент людей с хроническими заболеваниями:")
print(chronic_disease_rate)

obese_rate = chronic_disease_rate['Obese']
normal_rate = chronic_disease_rate['Normal']

print(f"Отношение: {obese_rate / normal_rate:.2f} раз")
````

**да, правда. хронические заболевания всречаются в полтора раза чаще**

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
df_filtered = df[df['Income Level'].isin(['High', 'Low'])].copy()
activity_mapping = {'Low': 1, 'Moderate': 2, 'High': 3}
df_filtered['Activity_Score'] = df_filtered['Physical Activity Level'].map(activity_mapping)

mean_activity = df_filtered.groupby('Income Level')['Activity_Score'].mean()
print(f"Низкий доход:  {mean_activity['Low']:.2f}")
print(f"Высокий доход: {mean_activity['High']:.2f}")
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
df_men = df[df['Gender'] == 'Male'].copy()

sun_median = df_men['Sun Exposure'].median()
df_high_sun = df_men[df_men['Sun Exposure'] > sun_median].copy()

high_income_count = (df_high_sun['Income Level'] == 'High').sum()
total = len(df_high_sun)

percent = high_income_count / total

print(f'Доля мужчин с высокой частотой пребывания на солнце и высоким заработком: {percent}')
````

1. Визуализируйте частоту встречаемых хронических заболеваний

2. Как связана оценка собственных когнитивных функций с уровнем стресса?

````python
x = df['Stress Levels']
y = df['Cognitive Function']

plt.figure(figsize=(8, 6))

plt.scatter(x, y, alpha=0.4, color='steelblue', label='Данные')

slope, intercept = np.polyfit(x, y, 1)
trendline = slope * x + intercept

plt.title('Связь уровня стресса и оценки когнитивных функций', fontsize=14, fontweight='bold')
plt.xlabel('Уровень стресса', fontsize=12)
plt.ylabel('Оценка когнитивных функций', fontsize=12)
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
````

4. Постройте box plot отображающий распределение уровня стресса у людей с различным уровнем дохода

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
counts = df.groupby(['Mental Health Status', 'Sleep Patterns']).size().unstack(fill_value=0)
mental_health_categories = ['Poor', 'Fair', 'Good', 'Excellent']
sleep_categories = ['Normal', 'Insomnia', 'Excessive']
colors = {'Normal': '#66b3ff', 'Insomnia': '#ff9999', 'Excessive': '#99ff99'}

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for idx, category in enumerate(mental_health_categories):
    data = counts.loc[category, sleep_categories]
    axes[idx].pie(data.values, 
                    labels=data.index, 
                    autopct='%1.1f%%',          
                    colors=[colors.get(x, '#cccccc') for x in data.index],
                    startangle=90)
    axes[idx].set_title(f'Ментальное здоровье: {category}', fontsize=10, fontweight='bold')
    axes[idx].axis('equal')

plt.suptitle("Паттерны сна по ментальному здоровью", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````
